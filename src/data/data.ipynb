```python
import time
from rdkit import Chem
from rdkit.Chem import AllChem
from openff.toolkit.topology import Molecule
from openff.toolkit.utils.toolkits import AmberToolsToolkitWrapper
import tempfile
import subprocess
import os
```

```python
from rdkit import Chem
mol = Chem.MolFromPDBFile("6i1e.pdb", removeHs=False)

for atom in mol.GetAtoms():
    info = atom.GetPDBResidueInfo()
    if info:
        print(f"Atom: {atom.GetSymbol()} | Residue: {info.GetResidueName()} {info.GetResidueNumber()} {info.GetChainId()}")
```

```python
from collections import defaultdict

residues = defaultdict(list)

for atom in mol.GetAtoms():
    info = atom.GetPDBResidueInfo()
    if info:
        key = (info.GetChainId(), info.GetResidueNumber(), info.GetResidueName())
        residues[key].append(atom)

for res_key, atoms in residues.items():
    print(f"Residue {res_key}: {len(atoms)} atoms")
```
